```python
import json
import pandas as pd

TARGET_CATS = {'cs.IR', 'cs.AI', 'cs.LG', 'cs.CL', 'cs.CV'}
MAX_PAPERS  = 100_000

records = []
path = '/kaggle/input/datasets/organizations/Cornell-University/arxiv/arxiv-metadata-oai-snapshot.json'

with open(path, 'r') as f:
    for line in f:
        paper     = json.loads(line)
        cats      = set(paper['categories'].split())
        year      = paper['update_date'][:4]

        if cats.intersection(TARGET_CATS) and year >= '2018':
            records.append({
                'id':         paper['id'],
                'title':      paper['title'].replace('\n', ' ').strip(),
                'abstract':   paper['abstract'].replace('\n', ' ').strip(),
                'categories': paper['categories'],
                'year':       year,
                'text':       paper['title'] + ' ' + paper['abstract'],
            })

        if len(records) >= MAX_PAPERS:
            break

df = pd.DataFrame(records)
print(f"Số paper lọc được: {len(df)}")
print(df['categories'].str.split().explode().value_counts().head(10))
df.head(3)
```

```python
print("=== Tổng quan dataset ===")
print(f"Số paper:     {len(df):,}")
print(f"Null title:   {df['title'].isna().sum()}")
print(f"Null abstract:{df['abstract'].isna().sum()}")
print(f"Năm cũ nhất:  {df['year'].min()}")
print(f"Năm mới nhất: {df['year'].max()}")
print(f"\nPhân bố theo năm:")
print(df['year'].value_counts().sort_index())
```

```python
# Lưu CSV sạch — dùng chung cho cả nhóm
df.to_csv('/kaggle/working/arxiv_cs_clean.csv', index=False)
print("Đã lưu! Vào tab Output bên phải để download.")
```
